```python
#split
import os
import random
from shutil import copyfile
from PIL import Image
import imagehash
import os
```

```python
def split_dataset(original_folder, train_folder, val_folder, test_folder, split_ratio=(0.8, 0.1, 0.1)):
    for snake_folder in os.listdir(original_folder):
        snake_folder_path = os.path.join(original_folder, snake_folder)
        train_snake_folder_path = os.path.join(train_folder, snake_folder)
        val_snake_folder_path = os.path.join(val_folder, snake_folder)
        test_snake_folder_path = os.path.join(test_folder, snake_folder)
        
        if os.path.isdir(snake_folder_path):
            os.makedirs(train_snake_folder_path, exist_ok=True)
            os.makedirs(val_snake_folder_path, exist_ok=True)
            os.makedirs(test_snake_folder_path, exist_ok=True)

            images = [f for f in os.listdir(snake_folder_path) if os.path.isfile(os.path.join(snake_folder_path, f))]
            random.shuffle(images)

            # Split images into train, val, and test sets
            total_images = len(images)
            train_size = int(split_ratio[0] * total_images)
            val_size = int(split_ratio[1] * total_images)

            train_images = images[:train_size]
            val_images = images[train_size:(train_size + val_size)]
            test_images = images[(train_size + val_size):]

            # Copy images to respective folders
            for img_filename in train_images:
                src_path = os.path.join(snake_folder_path, img_filename)
                dst_path = os.path.join(train_snake_folder_path, img_filename)
                copyfile(src_path, dst_path)

            for img_filename in val_images:
                src_path = os.path.join(snake_folder_path, img_filename)
                dst_path = os.path.join(val_snake_folder_path, img_filename)
                copyfile(src_path, dst_path)

            for img_filename in test_images:
                src_path = os.path.join(snake_folder_path, img_filename)
                dst_path = os.path.join(test_snake_folder_path, img_filename)
                copyfile(src_path, dst_path)

# Example usage:
original_dataset_folder = r"D:\NewDataset"
train_folder = r"E:\NewDataset\train"
val_folder = r"E:\NewDataset\val"
test_folder = r"E:\NewDataset\test"


# Create train, val, and test folders
os.makedirs(train_folder, exist_ok=True)
os.makedirs(val_folder, exist_ok=True)
os.makedirs(test_folder, exist_ok=True)


# Split the original dataset
split_dataset(original_dataset_folder, train_folder, val_folder, test_folder, split_ratio=(0.8, 0.1, 0.1))
```

```python


def augment_and_save_images(folder, output_folder):
    for snake_folder in os.listdir(folder):
        snake_folder_path = os.path.join(folder, snake_folder)
        output_snake_folder_path = os.path.join(output_folder, snake_folder)
        
        if os.path.isdir(snake_folder_path):
            os.makedirs(output_snake_folder_path, exist_ok=True)

            for filename in os.listdir(snake_folder_path):
                path = os.path.join(snake_folder_path, filename)
                with Image.open(path) as img:
                    # Original image
                    img.save(os.path.join(output_snake_folder_path, "original_" + filename))

                    # Rotate images without adding padding
                    for angle in [90, 180, 270]:
                        rotated_img = img.rotate(angle, expand=True)
                        rotated_img.save(os.path.join(output_snake_folder_path, f"rotate_{angle}_" + filename))

# Example usage:
dataset_folder = r"D:\Data Fyp\NewDataset"
rotated_dataset_folder = r"D:\Data Fyp\NewDataset"

# Create a rotated dataset folder
os.makedirs(rotated_dataset_folder, exist_ok=True)

# Augment and save rotated images without padding
augment_and_save_images(dataset_folder, rotated_dataset_folder)
```

```python
# duplicate remove


def dhash(image, hash_size=8):
    return imagehash.dhash(image, hash_size=hash_size)

def find_and_remove_duplicates(folder_path):
    # Dictionary to store image hashes
    hash_dict = {}

    for filename in os.listdir(folder_path):
        if filename.endswith(".jpg") or filename.endswith(".png"):
            file_path = os.path.join(folder_path, filename)

            # Calculate hash
            with Image.open(file_path) as img:
                img_hash = dhash(img)

            # Check if hash already exists
            if img_hash in hash_dict:
                print(f"Duplicate found in {folder_path}: {filename}. Deleting...")
                os.remove(file_path)
            else:
                hash_dict[img_hash] = file_path

# Replace "path/to/your/dataset" with the actual path to your dataset
dataset_path = r"C:\Users\Sandusha\FYP\Dataset\Val"

# Iterate over each folder in the dataset
for folder_name in os.listdir(dataset_path):
    folder_path = os.path.join(dataset_path, folder_name)

    # Check if it's a directory
    if os.path.isdir(folder_path):
        print(f"Checking for duplicates in folder: {folder_name}")
        find_and_remove_duplicates(folder_path)

print("Duplicate removal completed.")
```

```python
#mirror image
from PIL import Image, ImageOps

def augment_and_save_images(folder, output_folder):
    for snake_folder in os.listdir(folder):
        snake_folder_path = os.path.join(folder, snake_folder)
        output_snake_folder_path = os.path.join(output_folder, snake_folder)
        
        if os.path.isdir(snake_folder_path):
            os.makedirs(output_snake_folder_path, exist_ok=True)

            for filename in os.listdir(snake_folder_path):
                path = os.path.join(snake_folder_path, filename)
                with Image.open(path) as img:
                    # Original image
                    img.save(os.path.join(output_snake_folder_path, "original_" + filename))

                    # Mirror image
                    mirror_img = ImageOps.mirror(img)
                    mirror_img.save(os.path.join(output_snake_folder_path, "mirror_" + filename))

# Example usage:
dataset_folder = r"D:\Data Fyp\Dataset"
mirrored_dataset_folder = r"D:\Data Fyp\NewDataset"

# Create a mirrored dataset folder
os.makedirs(mirrored_dataset_folder, exist_ok=True)

# Augment and save mirrored images
augment_and_save_images(dataset_folder, mirrored_dataset_folder)
```

```python
#padd

def add_padding_to_images(folder, padding_color=(0, 0, 0)):
    for snake_folder in os.listdir(folder):
        snake_folder_path = os.path.join(folder, snake_folder)
        if os.path.isdir(snake_folder_path):
            for filename in os.listdir(snake_folder_path):
                path = os.path.join(snake_folder_path, filename)
                with Image.open(path) as img:
                    original_size = img.size
                    max_size = max(original_size)
                    new_size = (max_size, max_size)
                    
                    # Create a new image with the desired size and fill it with the padding color
                    padded_img = Image.new("RGB", new_size, padding_color)
                    
                    # Paste the original image onto the center of the new image
                    padded_img.paste(img, ((new_size[0] - original_size[0]) // 2, (new_size[1] - original_size[1]) // 2))
                    
                    # Save the padded image back to the original path
                    padded_img.save(path)

# Example usage:
dataset_folder = "D:\\Dataset2\\unpad"
add_padding_to_images(dataset_folder)
```

```python
#unpaddef unpad_images(folder):
    for snake_folder in os.listdir(folder):
        snake_folder_path = os.path.join(folder, snake_folder)
        if os.path.isdir(snake_folder_path):
            for filename in os.listdir(snake_folder_path):
                path = os.path.join(snake_folder_path, filename)
                with Image.open(path) as img:
                    # Find the non-zero region in the image
                    non_zero_region = img.getbbox()
                    
                    # Crop the image to the non-zero region
                    cropped_img = img.crop(non_zero_region)
                    
                    # Save the cropped image back to the original path
                    cropped_img.save(path)

dataset_folder = "D:\\Dataset2\\unpad"
unpad_images(dataset_folder)
# Unpad the images
```
