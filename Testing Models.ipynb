```python
import os
from PIL import Image
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report, f1_score, precision_score, recall_score
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.preprocessing import image

# Load the trained model
model = load_model('mobilenet-ft.model')
print("Prediction result:")

# Directory containing the test dataset
test_dir = "D:\\DatasetNew\\test"
classes = sorted(os.listdir(test_dir))
IM_WIDTH, IM_HEIGHT = 224, 224

true_labels = []
predicted_labels = []

class_labels = {i: class_name for i, class_name in enumerate(classes)}

# Function to preprocess image for validation
def preprocess_image(img_path):
    img = Image.open(img_path).convert('RGB')
    img = img.resize((IM_WIDTH, IM_HEIGHT))
    img_array = np.expand_dims(np.array(img), axis=0)
    img_array = img_array / 255.0  # Rescale to [0, 1]
    return img_array

# Loop through each class and each image in the test set
for i, class_name in enumerate(classes):
    class_path = os.path.join(test_dir, class_name)
    files = [file for file in os.listdir(class_path) if file.lower().endswith(('.png', '.jpg', '.jpeg'))]

    for file_name in files:
        image_path = os.path.join(class_path, file_name)
        if os.path.isfile(image_path):
            # Preprocess and predict the image
            img_array = preprocess_image(image_path)
            predictions = model.predict(img_array)
            predicted_label = np.argmax(predictions, axis=1)[0]
            predicted_class = class_labels[predicted_label]

            true_labels.append(class_name)
            predicted_labels.append(predicted_class)

            # Print each prediction result
            print(f"File: {file_name}, Predicted: {predicted_class}, Actual: {class_name}")
        else:
            print(f"Skipping {file_name} in {class_name} as it is not a file.")

# Ensure there are predictions
if len(true_labels) > 0:
    # Confusion Matrix
    cm = confusion_matrix(true_labels, predicted_labels, labels=classes)
    
    # Plot Confusion Matrix with seaborn for styling
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
    plt.title("Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

    # Classification Report
    class_report = classification_report(true_labels, predicted_labels, target_names=classes)
    print("\nClassification Report:")
    print(class_report)

    # F1 Score, Precision, and Recall
    f1_score_value = f1_score(true_labels, predicted_labels, average='weighted')
    precision_value = precision_score(true_labels, predicted_labels, average='weighted')
    recall_value = recall_score(true_labels, predicted_labels, average='weighted')

    print(f"\nF1 Score: {f1_score_value:.4f}")
    print(f"Precision: {precision_value:.4f}")
    print(f"Recall: {recall_value:.4f}")
else:
    print("No valid test files found.")
```
