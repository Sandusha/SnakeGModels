```python
#matrix for mobilenet model

import os
from PIL import Image
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

print("Prediction result:")
test_dir = "Dataset/Test"  # Replace with the actual path to your test directory
classes = sorted(os.listdir(test_dir))

correct = 0
total = 0
true_labels = []
predicted_labels = []

# Define the class labels based on your dataset
class_labels = {i: class_name for i, class_name in enumerate(classes)}

for i, class_name in enumerate(classes):
    class_path = os.path.join(test_dir, class_name)
    files = [file for file in os.listdir(class_path) if file.lower().endswith(('.png', '.jpg', '.jpeg'))]

    for file_name in files:
        total += 1
        image_path = os.path.join(class_path, file_name)

        # Check if the path is a file
        if os.path.isfile(image_path):
            # Load and preprocess the image
            image = Image.open(image_path).convert('RGB')
            image = image.resize((IM_WIDTH, IM_HEIGHT))
            image = np.expand_dims(image, axis=0)
            image = np.array(image)
            image = image / 255.0

            # Predict the class label
            predictions = model.predict([image])
            pred = np.argmax(predictions, axis=1)
            predicted_class = class_labels[pred[0]]

            # Store true and predicted labels
            true_labels.append(class_name)
            predicted_labels.append(predicted_class)

            # Print the result
            print(f"{total}. {file_name}: Predicted - {predicted_class}, Actual - {class_name}")

            # Check correctness
            if class_name.lower() == predicted_class.lower():
                correct += 1
        else:
            print(f"Skipping {file_name} in {class_name} as it is not a file.")

# Check if any files were processed to avoid division by zero
if total > 0:
    # Calculate accuracy
    accuracy = correct / total
    print(f"Accuracy: {accuracy}")

    # Calculate confusion matrix
    cm = confusion_matrix(true_labels, predicted_labels, labels=classes)
    print("\nConfusion Matrix:")
    print(cm)

    # Print classification report
    print("\nClassification Report:")
    print(classification_report(true_labels, predicted_labels, target_names=classes))
else:
    print("No files found in the specified directory.")
```
